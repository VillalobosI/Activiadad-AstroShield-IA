```python
# Archivo: algoritmo_evolutivo.py
# Representación de los 8 satélites y estructura del algoritmo evolutivo
import random
import copy
import restricciones_fitness as rf
NUM_SATELITES = 8
def crear_satelite_aleatorio():
    """Crea un satélite con latitud [-90, 90] y longitud [-180, 180] aleatorias."""
    latitud = random.uniform(-90.0, 90.0)
    longitud = random.uniform(-180.0, 180.0)
    return [latitud, longitud]
def crear_configuracion_aleatoria():
    """Genera una constelación de 8 satélites en posiciones aleatorias."""
    configuracion = []
    for i in range(NUM_SATELITES):
        satelite = crear_satelite_aleatorio()
        configuracion.append(satelite)
    return configuracion
def crear_poblacion_inicial(tamano_poblacion):
    """Crea una lista de configuraciones aleatorias."""
    poblacion = []
    for i in range(tamano_poblacion):
        individuo = crear_configuracion_aleatoria()
        poblacion.append(individuo)
    return poblacion
def seleccion_por_torneo(poblacion, puntajes_fitness, tamano_torneo=3):
    """Selecciona al mejor individuo de una muestra aleatoria de la población."""
    indices_participantes = []
    total_individuos = len(poblacion)
    for i in range(tamano_torneo):
        indice_azar = random.randint(0, total_individuos - 1)
        indices_participantes.append(indice_azar)
    mejor_indice = indices_participantes[0]
    mejor_puntaje = puntajes_fitness[mejor_indice]
    for indice in indices_participantes:
        if puntajes_fitness[indice] > mejor_puntaje:
            mejor_puntaje = puntajes_fitness[indice]
            mejor_indice = indice
    return copy.deepcopy(poblacion[mejor_indice])
def cruzar_configuraciones(padre1, padre2):
    """Cruza dos configuraciones de 8 satélites usando un punto de corte."""
    punto_corte = random.randint(1, NUM_SATELITES - 1)
    hijo = []
    for i in range(NUM_SATELITES):
        if i < punto_corte:
            hijo.append(copy.deepcopy(padre1[i]))
        else:
            hijo.append(copy.deepcopy(padre2[i]))
    return hijo
def mutar_configuracion(configuracion, probabilidad_mutacion=0.25):
    """Modifica ligeramente las coordenadas de los satélites según una probabilidad."""
    nueva_config = copy.deepcopy(configuracion)
    for i in range(NUM_SATELITES):
        azar = random.random()
        if azar < probabilidad_mutacion:
            # 80% de las veces hacemos un ajuste pequeño (búsqueda local)
            # 20% de las veces movemos el satélite a una posición totalmente nueva
            if random.random() < 0.8:
                cambio_lat = random.gauss(0, 8.0)
                cambio_lon = random.gauss(0, 8.0)
                nueva_lat = nueva_config[i][0] + cambio_lat
                nueva_lon = nueva_config[i][1] + cambio_lon
                # Mantener latitud entre -90 y 90
                if nueva_lat > 90.0:
                    nueva_lat = 90.0
                if nueva_lat < -90.0:
                    nueva_lat = -90.0
                # Mantener longitud entre -180 y 180
                if nueva_lon > 180.0:
                    nueva_lon = nueva_lon - 360.0
                if nueva_lon < -180.0:
                    nueva_lon = nueva_lon + 360.0
                nueva_config[i] = [nueva_lat, nueva_lon]
            else:
                nueva_config[i] = crear_satelite_aleatorio()
    return nueva_config
def ejecutar_algoritmo_evolutivo(
    tamano_poblacion=60,
    generaciones=100,
    probabilidad_mutacion=0.25,
    regiones=None,
    zonas_desechos=None,
    objetos_ocupados=None,
    mostrar_progreso=True
):
    """
    Ejecuta el ciclo principal del algoritmo evolutivo y devuelve:
    - mejor_configuracion_inicial
    - mejor_configuracion_final
    - historial_mejor_fitness
    - historial_promedio_fitness
    """
    if regiones is None:
        regiones = rf.obtener_regiones_iniciales()
    if zonas_desechos is None:
        zonas_desechos = rf.obtener_zonas_desechos_iniciales()
    if objetos_ocupados is None:
        objetos_ocupados = rf.obtener_objetos_ocupados_iniciales()
    poblacion = crear_poblacion_inicial(tamano_poblacion)
    historial_mejor_fitness = []
    historial_promedio_fitness = []
    mejor_configuracion_inicial = None
    mejor_configuracion_global = None
    mejor_fitness_global = -999999.0
    for gen in range(generaciones):
        puntajes_fitness = []
        suma_fitness = 0.0
        # 1. Evaluar toda la población actual
        for individuo in poblacion:
            resultado = rf.evaluar_configuracion(
                individuo, regiones, zonas_desechos, objetos_ocupados
            )
            fitness_ind = resultado["fitness"]
            puntajes_fitness.append(fitness_ind)
            suma_fitness += fitness_ind
            if fitness_ind > mejor_fitness_global:
                mejor_fitness_global = fitness_ind
                mejor_configuracion_global = copy.deepcopy(individuo)
        # Guardar la mejor configuración de la generación 0 para comparar después
        if gen == 0:
            indice_mejor_gen0 = 0
            for i in range(len(puntajes_fitness)):
                if puntajes_fitness[i] > puntajes_fitness[indice_mejor_gen0]:
                    indice_mejor_gen0 = i
            mejor_configuracion_inicial = copy.deepcopy(poblacion[indice_mejor_gen0])
        promedio_actual = suma_fitness / tamano_poblacion
        historial_mejor_fitness.append(mejor_fitness_global)
        historial_promedio_fitness.append(promedio_actual)
        if mostrar_progreso and (gen % 20 == 0 or gen == generaciones - 1):
            info_mejor = rf.evaluar_configuracion(
                mejor_configuracion_global, regiones, zonas_desechos, objetos_ocupados
            )
            print(
                f"Generación {gen:3d} | Mejor Fitness: {mejor_fitness_global:7.2f} | "
                f"Promedio: {promedio_actual:7.2f} | "
                f"Regiones cubiertas: {len(info_mejor['regiones_cubiertas'])}/{len(regiones)} | "
                f"Válida: {info_mejor['es_valida']}"
            )
        # 2. Ordenar índices de mejor a peor para aplicar Elitismo (conservar los 2 mejores)
        indices_ordenados = list(range(tamano_poblacion))
        for i in range(tamano_poblacion):
            for j in range(i + 1, tamano_poblacion):
                if puntajes_fitness[indices_ordenados[j]] > puntajes_fitness[indices_ordenados[i]]:
                    temp = indices_ordenados[i]
                    indices_ordenados[i] = indices_ordenados[j]
                    indices_ordenados[j] = temp
        nueva_poblacion = []
        nueva_poblacion.append(copy.deepcopy(poblacion[indices_ordenados[0]]))
        nueva_poblacion.append(copy.deepcopy(poblacion[indices_ordenados[1]]))
        # 3. Generar el resto de la nueva población mediante torneo, cruza y mutación
        while len(nueva_poblacion) < tamano_poblacion:
            padre1 = seleccion_por_torneo(poblacion, puntajes_fitness)
            padre2 = seleccion_por_torneo(poblacion, puntajes_fitness)
            hijo = cruzar_configuraciones(padre1, padre2)
            hijo_mutado = mutar_configuracion(hijo, probabilidad_mutacion)
            nueva_poblacion.append(hijo_mutado)
        poblacion = nueva_poblacion
    return {
        "configuracion_inicial": mejor_configuracion_inicial,
        "configuracion_final": mejor_configuracion_global,
        "historial_mejor": historial_mejor_fitness,
        "historial_promedio": historial_promedio_fitness
    }
if __name__ == "__main__":
    print("=== Ejecutando Algoritmo Evolutivo AstroShield ===")
    resultados = ejecutar_algoritmo_evolutivo(
        tamano_poblacion=60,
        generaciones=100,
        probabilidad_mutacion=0.25,
        mostrar_progreso=True
    )
    evaluacion_final = rf.evaluar_configuracion(resultados["configuracion_final"])
    print("\n=== Resumen de la Configuración Final ===")
    print("Fitness Final:", round(evaluacion_final["fitness"], 2))
    print("¿Cumple todas las restricciones?:", evaluacion_final["es_valida"])
    print("Regiones cubiertas:", evaluacion_final["regiones_cubiertas"])
    print("Regiones NO cubiertas:", evaluacion_final["regiones_no_cubiertas"])
    print("\nCoordenadas de los 8 satélites (Latitud, Longitud):")
    for idx, sat in enumerate(resultados["configuracion_final"]):
        print(f"  Satélite {idx + 1}: Lat = {sat[0]:6.2f}°, Lon = {sat[1]:7.2f}°")
```
