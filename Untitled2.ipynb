```python
# PROYECTO: AstroShield — Evolucionar una constelación inteligente
import math
import random
import copy
import numpy as np
import matplotlib.pyplot as plt

#==============================================================================
# BLOQUE 3 (COMMIT 3): restricciones_fitness.py
# Contiene la función de fitness, el cálculo de cobertura de las 6 regiones
# con sus prioridades, y el filtro de las 3 zonas de desechos y 2 objetos
#Elaborado por Iñaki Villalobos
# ==============================================================================

RADIO_TIERRA_KM = 6371.0
RADIO_OBSERVACION_REGION_KM = 300.0
ALCANCE_MAXIMO_SATELITE_KM = 1200.0  # Radio dentro del cual el satélite logra cubrir los 300 km de la región


def obtener_regiones_iniciales():
    """Devuelve las 6 regiones de interés con sus coordenadas y prioridades."""
    return [
        {"nombre": "Península de Yucatán, México", "lat": 20.5, "lon": -89.0, "prioridad": "Alta", "puntos": 300},
        {"nombre": "Amazonía brasileña", "lat": -3.5, "lon": -62.0, "prioridad": "Alta", "puntos": 300},
        {"nombre": "California, Estados Unidos", "lat": 36.8, "lon": -119.4, "prioridad": "Media", "puntos": 200},
        {"nombre": "Región del Sahel, África", "lat": 15.0, "lon": 0.0, "prioridad": "Media", "puntos": 200},
        {"nombre": "Japón", "lat": 36.0, "lon": 138.0, "prioridad": "Media", "puntos": 200},
        {"nombre": "Interior de Australia", "lat": -25.0, "lon": 133.0, "prioridad": "Normal", "puntos": 100}
    ]


def obtener_zonas_desechos_iniciales():
    """Devuelve las 3 zonas restringidas por desechos espaciales (radios en grados)."""
    return [
        {"nombre": "Atlántico Norte", "lat": 45.0, "lon": -35.0, "radio_grados": 12.0},
        {"nombre": "Pacífico Sur", "lat": -30.0, "lon": -120.0, "radio_grados": 10.0},
        {"nombre": "Región polar norte", "lat": 78.0, "lon": 30.0, "radio_grados": 8.0}
    ]


def obtener_objetos_ocupados_iniciales():
    """Devuelve los 2 objetos existentes con su radio de exclusión en grados."""
    return [
        {"nombre": "Objeto A", "lat": 10.0, "lon": 45.0, "radio_grados": 5.0},
        {"nombre": "Objeto B", "lat": -15.0, "lon": 110.0, "radio_grados": 5.0}
    ]


def calcular_distancia_angular_grados(lat1, lon1, lat2, lon2):
    """
    Calcula la distancia angular en grados sobre una esfera entre dos coordenadas
    usando la fórmula esférica de Haversine.
    """
    rad_lat1 = math.radians(lat1)
    rad_lon1 = math.radians(lon1)
    rad_lat2 = math.radians(lat2)
    rad_lon2 = math.radians(lon2)

    delta_lat = rad_lat2 - rad_lat1
    delta_lon = rad_lon2 - rad_lon1

    a = (math.sin(delta_lat / 2.0) ** 2) + \
        math.cos(rad_lat1) * math.cos(rad_lat2) * (math.sin(delta_lon / 2.0) ** 2)

    if a > 1.0:
        a = 1.0
    if a < 0.0:
        a = 0.0

    angulo_central_rad = 2.0 * math.atan2(math.sqrt(a), math.sqrt(1.0 - a))
    distancia_grados = math.degrees(angulo_central_rad)
    return distancia_grados


def calcular_distancia_km(lat1, lon1, lat2, lon2):
    """Calcula la distancia superficial en kilómetros entre dos puntos de la Tierra."""
    grados = calcular_distancia_angular_grados(lat1, lon1, lat2, lon2)
    radianes = math.radians(grados)
    return RADIO_TIERRA_KM * radianes


def verificar_restricciones(configuracion, zonas_desechos, objetos_ocupados):
    """
    Revisa si algún satélite invade las 3 zonas de desechos, los 2 objetos ocupados,
    o si dos satélites chocan entre sí.
    """
    violaciones = []

    for i in range(len(configuracion)):
        sat_lat = configuracion[i][0]
        sat_lon = configuracion[i][1]

        # 1. Revisar las 3 zonas de desechos espaciales
        for zona in zonas_desechos:
            dist_grados = calcular_distancia_angular_grados(
                sat_lat, sat_lon, zona["lat"], zona["lon"]
            )
            if dist_grados <= zona["radio_grados"]:
                violaciones.append(
                    f"Satélite {i + 1} invade zona de desechos '{zona['nombre']}' "
                    f"(distancia {dist_grados:.2f}° <= {zona['radio_grados']}°)"
                )

        # 2. Revisar los 2 objetos existentes
        for obj in objetos_ocupados:
            dist_grados = calcular_distancia_angular_grados(
                sat_lat, sat_lon, obj["lat"], obj["lon"]
            )
            if dist_grados <= obj["radio_grados"]:
                violaciones.append(
                    f"Satélite {i + 1} invade zona de exclusión de '{obj['nombre']}' "
                    f"(distancia {dist_grados:.2f}° <= {obj['radio_grados']}°)"
                )

        # 3. Revisar separación mínima entre satélites de la misma constelación (mínimo 2°)
        for j in range(i + 1, len(configuracion)):
            dist_entre_sat = calcular_distancia_angular_grados(
                sat_lat, sat_lon, configuracion[j][0], configuracion[j][1]
            )
            if dist_entre_sat < 2.0:
                violaciones.append(
                    f"Satélites {i + 1} y {j + 1} demasiado cerca ({dist_entre_sat:.2f}° < 2.0°)"
                )

    return violaciones


def evaluar_configuracion(configuracion, regiones=None, zonas_desechos=None, objetos_ocupados=None):
    """
    Calcula el fitness total de una constelación de 8 satélites, considerando:
    - Puntos por cubrir cada una de las 6 regiones según su prioridad.
    - Bono por cercanía óptima dentro del radio de 300 km de cada región.
    - Penalización fuerte por cada violación a zonas de desechos u objetos existentes.
    """
    if regiones is None:
        regiones = obtener_regiones_iniciales()
    if zonas_desechos is None:
        zonas_desechos = obtener_zonas_desechos_iniciales()
    if objetos_ocupados is None:
        objetos_ocupados = obtener_objetos_ocupados_iniciales()

    puntaje_total = 0.0
    regiones_cubiertas = []
    regiones_no_cubiertas = []
    distancias_minimas_regiones = {}

    # 1. Evaluar cobertura de las 6 regiones
    for region in regiones:
        mejor_distancia_km = 999999.0

        for satelite in configuracion:
            dist_km = calcular_distancia_km(
                satelite[0], satelite[1], region["lat"], region["lon"]
            )
            if dist_km < mejor_distancia_km:
                mejor_distancia_km = dist_km

        distancias_minimas_regiones[region["nombre"]] = round(mejor_distancia_km, 2)

        if mejor_distancia_km <= ALCANCE_MAXIMO_SATELITE_KM:
            regiones_cubiertas.append(region["nombre"])
            puntaje_total += region["puntos"]

            # Bono extra si el satélite se posiciona dentro de los 300 km centrales de observación
            if mejor_distancia_km <= RADIO_OBSERVACION_REGION_KM:
                bono_precision = 50.0 * (1.0 - (mejor_distancia_km / RADIO_OBSERVACION_REGION_KM))
                puntaje_total += bono_precision
            else:
                proporcion = (ALCANCE_MAXIMO_SATELITE_KM - mejor_distancia_km) / (
                    ALCANCE_MAXIMO_SATELITE_KM - RADIO_OBSERVACION_REGION_KM
                )
                puntaje_total += 20.0 * proporcion
        else:
            regiones_no_cubiertas.append(region["nombre"])
            guia_cercania = max(0.0, 20.0 * (1.0 - (mejor_distancia_km / 10000.0)))
            puntaje_total += guia_cercania

    # 2. Bono por buena distribución entre satélites (evitar amontonamiento innecesario)
    for i in range(len(configuracion)):
        for j in range(i + 1, len(configuracion)):
            dist_grados = calcular_distancia_angular_grados(
                configuracion[i][0], configuracion[i][1],
                configuracion[j][0], configuracion[j][1]
            )
            if dist_grados >= 10.0:
                puntaje_total += 1.5

    # 3. Aplicar filtro de restricciones (zonas de desechos y posiciones ocupadas)
    violaciones = verificar_restricciones(configuracion, zonas_desechos, objetos_ocupados)
    es_valida = len(violaciones) == 0

    if not es_valida:
        puntaje_total -= 1000.0 * len(violaciones)

    return {
        "fitness": puntaje_total,
        "es_valida": es_valida,
        "violaciones": violaciones,
        "regiones_cubiertas": regiones_cubiertas,
        "regiones_no_cubiertas": regiones_no_cubiertas,
        "distancias_km": distancias_minimas_regiones
    }

# ==============================================================================
# BLOQUE 2 (Venus): algoritmo_evolutivo.py
# Contiene la representación de los 8 satélites y la estructura principal
# del algoritmo evolutivo (generación, selección por torneo, cruza y mutación).
# ==============================================================================
NUM_SATELITES = 8
def crear_satelite_aleatorio():
    """Crea un satélite con latitud [-90, 90] y longitud [-180, 180] aleatorias."""
    latitud = random.uniform(-90.0, 90.0)
    longitud = random.uniform(-180.0, 180.0)
    return [latitud, longitud]
def crear_configuracion_aleatoria():
    """Genera una constelación de 8 satélites en posiciones aleatorias."""
    configuracion = []
    for i in range(NUM_SATELITES):
        satelite = crear_satelite_aleatorio()
        configuracion.append(satelite)
    return configuracion
def crear_poblacion_inicial(tamano_poblacion):
    """Crea una lista de configuraciones aleatorias."""
    poblacion = []
    for i in range(tamano_poblacion):
        individuo = crear_configuracion_aleatoria()
        poblacion.append(individuo)
    return poblacion
def seleccion_por_torneo(poblacion, puntajes_fitness, tamano_torneo=3):
    """Selecciona al mejor individuo de una muestra aleatoria de la población."""
    indices_participantes = []
    total_individuos = len(poblacion)
    for i in range(tamano_torneo):
        indice_azar = random.randint(0, total_individuos - 1)
        indices_participantes.append(indice_azar)
    mejor_indice = indices_participantes[0]
    mejor_puntaje = puntajes_fitness[mejor_indice]
    for indice in indices_participantes:
        if puntajes_fitness[indice] > mejor_puntaje:
            mejor_puntaje = puntajes_fitness[indice]
            mejor_indice = indice
    return copy.deepcopy(poblacion[mejor_indice])
def cruzar_configuraciones(padre1, padre2):
    """Cruza dos configuraciones de 8 satélites usando un punto de corte."""
    punto_corte = random.randint(1, NUM_SATELITES - 1)
    hijo = []
    for i in range(NUM_SATELITES):
        if i < punto_corte:
            hijo.append(copy.deepcopy(padre1[i]))
        else:
            hijo.append(copy.deepcopy(padre2[i]))
    return hijo
def mutar_configuracion(configuracion, probabilidad_mutacion=0.25):
    """Modifica ligeramente las coordenadas de los satélites según una probabilidad."""
    nueva_config = copy.deepcopy(configuracion)
    for i in range(NUM_SATELITES):
        azar = random.random()
        if azar < probabilidad_mutacion:
            # 80% de las veces hacemos un ajuste fino Gaussiano
            # 20% de las veces movemos el satélite a una posición totalmente nueva
            if random.random() < 0.8:
                cambio_lat = random.gauss(0, 8.0)
                cambio_lon = random.gauss(0, 8.0)
                nueva_lat = nueva_config[i][0] + cambio_lat
                nueva_lon = nueva_config[i][1] + cambio_lon
                if nueva_lat > 90.0:
                    nueva_lat = 90.0
                if nueva_lat < -90.0:
                    nueva_lat = -90.0
                if nueva_lon > 180.0:
                    nueva_lon = nueva_lon - 360.0
                if nueva_lon < -180.0:
                    nueva_lon = nueva_lon + 360.0
                nueva_config[i] = [nueva_lat, nueva_lon]
            else:
                nueva_config[i] = crear_satelite_aleatorio()
    return nueva_config
def ejecutar_algoritmo_evolutivo(
    tamano_poblacion=60,
    generaciones=100,
    probabilidad_mutacion=0.25,
    regiones=None,
    zonas_desechos=None,
    objetos_ocupados=None,
    mostrar_progreso=True
):
    """
    Ejecuta el ciclo principal del algoritmo evolutivo y devuelve:
    - configuracion_inicial (mejor de la generación 0)
    - configuracion_final (mejor global encontrada)
    - historial_mejor
    - historial_promedio
    """
    if regiones is None:
        regiones = obtener_regiones_iniciales()
    if zonas_desechos is None:
        zonas_desechos = obtener_zonas_desechos_iniciales()
    if objetos_ocupados is None:
        objetos_ocupados = obtener_objetos_ocupados_iniciales()
    poblacion = crear_poblacion_inicial(tamano_poblacion)
    historial_mejor_fitness = []
    historial_promedio_fitness = []
    mejor_configuracion_inicial = None
    mejor_configuracion_global = None
    mejor_fitness_global = -999999.0
    for gen in range(generaciones):
        puntajes_fitness = []
        suma_fitness = 0.0
        # 1. Evaluar toda la población actual
        for individuo in poblacion:
            resultado = evaluar_configuracion(
                individuo, regiones, zonas_desechos, objetos_ocupados
            )
            fitness_ind = resultado["fitness"]
            puntajes_fitness.append(fitness_ind)
            suma_fitness += fitness_ind
            if fitness_ind > mejor_fitness_global:
                mejor_fitness_global = fitness_ind
                mejor_configuracion_global = copy.deepcopy(individuo)
        # Guardar la mejor configuración de la generación 0 para comparar después
        if gen == 0:
            indice_mejor_gen0 = 0
            for i in range(len(puntajes_fitness)):
                if puntajes_fitness[i] > puntajes_fitness[indice_mejor_gen0]:
                    indice_mejor_gen0 = i
            mejor_configuracion_inicial = copy.deepcopy(poblacion[indice_mejor_gen0])
        promedio_actual = suma_fitness / tamano_poblacion
        historial_mejor_fitness.append(mejor_fitness_global)
        historial_promedio_fitness.append(promedio_actual)
        if mostrar_progreso and (gen % 20 == 0 or gen == generaciones - 1):
            info_mejor = evaluar_configuracion(
                mejor_configuracion_global, regiones, zonas_desechos, objetos_ocupados
            )
            print(
                f"Generación {gen:3d} | Mejor Fitness: {mejor_fitness_global:7.2f} | "
                f"Promedio: {promedio_actual:7.2f} | "
                f"Regiones cubiertas: {len(info_mejor['regiones_cubiertas'])}/{len(regiones)} | "
                f"Válida: {info_mejor['es_valida']}"
            )
        # 2. Ordenar índices de mejor a peor para aplicar Elitismo (conservar los 2 mejores)
        indices_ordenados = list(range(tamano_poblacion))
        for i in range(tamano_poblacion):
            for j in range(i + 1, tamano_poblacion):
                if puntajes_fitness[indices_ordenados[j]] > puntajes_fitness[indices_ordenados[i]]:
                    temp = indices_ordenados[i]
                    indices_ordenados[i] = indices_ordenados[j]
                    indices_ordenados[j] = temp
        nueva_poblacion = []
        nueva_poblacion.append(copy.deepcopy(poblacion[indices_ordenados[0]]))
        nueva_poblacion.append(copy.deepcopy(poblacion[indices_ordenados[1]]))
        # 3. Generar el resto de la nueva población mediante torneo, cruza y mutación
        while len(nueva_poblacion) < tamano_poblacion:
            padre1 = seleccion_por_torneo(poblacion, puntajes_fitness)
            padre2 = seleccion_por_torneo(poblacion, puntajes_fitness)
            hijo = cruzar_configuraciones(padre1, padre2)
            hijo_mutado = mutar_configuracion(hijo, probabilidad_mutacion)
            nueva_poblacion.append(hijo_mutado)
        poblacion = nueva_poblacion
    return {
        "configuracion_inicial": mejor_configuracion_inicial,
        "configuracion_final": mejor_configuracion_global,
        "historial_mejor": historial_mejor_fitness,
        "historial_promedio": historial_promedio_fitness
    }
```

```python
# ==============================================================================
# BLOQUE 4 (Angel): visualizacion_3d.py
# Genera el globo terrestre en 3D marcando los 8 satélites, las 6 regiones
# objetivo, las 3 zonas restringidas por desechos y los 2 objetos existentes.
# ==============================================================================
def convertir_lat_lon_a_xyz(lat_grados, lon_grados, radio=1.0):
    """Convierte coordenadas esféricas (latitud, longitud) a coordenadas cartesianas 3D (x, y, z)."""
    lat_rad = math.radians(lat_grados)
    lon_rad = math.radians(lon_grados)
    x = radio * math.cos(lat_rad) * math.cos(lon_rad)
    y = radio * math.cos(lat_rad) * math.sin(lon_rad)
    z = radio * math.sin(lat_rad)
    return x, y, z
def generar_circulo_esferico_xyz(lat_centro, lon_centro, radio_grados, radio_esfera=1.01, num_puntos=40):
    """
    Genera los puntos 3D de un círculo sobre la superficie del globo terrestre
    para representar visualmente los radios de restricción o de observación.
    """
    xs, ys, zs = [], [], []
    lat_c_rad = math.radians(lat_centro)
    lon_c_rad = math.radians(lon_centro)
    dist_ang_rad = math.radians(radio_grados)
    for i in range(num_puntos + 1):
        angulo_direccion = 2.0 * math.pi * i / num_puntos
        lat_p = math.asin(
            math.sin(lat_c_rad) * math.cos(dist_ang_rad) +
            math.cos(lat_c_rad) * math.sin(dist_ang_rad) * math.cos(angulo_direccion)
        )
        lon_p = lon_c_rad + math.atan2(
            math.sin(angulo_direccion) * math.sin(dist_ang_rad) * math.cos(lat_c_rad),
            math.cos(dist_ang_rad) - math.sin(lat_c_rad) * math.sin(lat_p)
        )
        x = radio_esfera * math.cos(lat_p) * math.cos(lon_p)
        y = radio_esfera * math.cos(lat_p) * math.sin(lon_p)
        z = radio_esfera * math.sin(lat_p)
        xs.append(x)
        ys.append(y)
        zs.append(z)
    return xs, ys, zs
def dibujar_escenario_en_subgrafico(ax, configuracion, titulo, regiones, zonas_desechos, objetos_ocupados):
    """Dibuja el globo terrestre 3D completo en un eje de Matplotlib."""
    # 1. Crear malla esférica de la Tierra
    u = np.linspace(0, 2 * np.pi, 36)
    v = np.linspace(0, np.pi, 18)
    x_tierra = np.outer(np.cos(u), np.sin(v))
    y_tierra = np.outer(np.sin(u), np.sin(v))
    z_tierra = np.outer(np.ones(np.size(u)), np.cos(v))
    ax.plot_surface(x_tierra, y_tierra, z_tierra, color="lightblue", alpha=0.35, edgecolor="steelblue", linewidth=0.3)
    resultado = evaluar_configuracion(configuracion, regiones, zonas_desechos, objetos_ocupados)
    cubiertas = resultado["regiones_cubiertas"]
    mostro_etiqueta_cubierta = False
    mostro_etiqueta_no_cubierta = False
    # 2. Dibujar las 6 regiones a observar (Verde = Cubierta, Naranja = No cubierta)
    for reg in regiones:
        esta_cubierta = reg["nombre"] in cubiertas
        color_reg = "limegreen" if esta_cubierta else "darkorange"
        etiqueta_reg = None
        if esta_cubierta and not mostro_etiqueta_cubierta:
            etiqueta_reg = "Región Cubierta"
            mostro_etiqueta_cubierta = True
        elif not esta_cubierta and not mostro_etiqueta_no_cubierta:
            etiqueta_reg = "Región NO Cubierta"
            mostro_etiqueta_no_cubierta = True
        rx, ry, rz = convertir_lat_lon_a_xyz(reg["lat"], reg["lon"], radio=1.02)
        ax.scatter([rx], [ry], [rz], color=color_reg, s=90, marker="o", edgecolors="black", label=etiqueta_reg)
        ax.text(rx * 1.06, ry * 1.06, rz * 1.06, reg["nombre"].split(",")[0], fontsize=7, color="darkgreen")
        # Dibujar radio de observación (~2.7° equivale a 300 km)
        cx, cy, cz = generar_circulo_esferico_xyz(reg["lat"], reg["lon"], 2.7, radio_esfera=1.01)
        ax.plot(cx, cy, cz, color=color_reg, linewidth=1.2)
    # 3. Dibujar las 3 zonas restringidas por desechos espaciales (Rojo)
    for idx, zona in enumerate(zonas_desechos):
        zx, zy, zz = convertir_lat_lon_a_xyz(zona["lat"], zona["lon"], radio=1.02)
        etiqueta_z = "Zona de Desechos" if idx == 0 else None
        ax.scatter([zx], [zy], [zz], color="red", s=80, marker="X", label=etiqueta_z)
        ax.text(zx * 1.06, zy * 1.06, zz * 1.06, zona["nombre"], fontsize=7, color="darkred")
        cx, cy, cz = generar_circulo_esferico_xyz(zona["lat"], zona["lon"], zona["radio_grados"], radio_esfera=1.02)
        ax.plot(cx, cy, cz, color="red", linestyle="--", linewidth=1.5)
    # 4. Dibujar las 2 posiciones ocupadas por objetos existentes (Morado)
    for idx, obj in enumerate(objetos_ocupados):
        ox, oy, oz = convertir_lat_lon_a_xyz(obj["lat"], obj["lon"], radio=1.02)
        etiqueta_o = "Objeto Existente (Ocupado)" if idx == 0 else None
        ax.scatter([ox], [oy], [oz], color="purple", s=80, marker="s", label=etiqueta_o)
        ax.text(ox * 1.06, oy * 1.06, oz * 1.06, obj["nombre"], fontsize=7, color="purple")
        cx, cy, cz = generar_circulo_esferico_xyz(obj["lat"], obj["lon"], obj["radio_grados"], radio_esfera=1.02)
        ax.plot(cx, cy, cz, color="purple", linestyle=":", linewidth=1.5)
    # 5. Dibujar los 8 satélites de la constelación en órbita (Azul brillante a radio=1.12)
    for idx, sat in enumerate(configuracion):
        sx, sy, sz = convertir_lat_lon_a_xyz(sat[0], sat[1], radio=1.12)
        tx, ty, tz = convertir_lat_lon_a_xyz(sat[0], sat[1], radio=1.00)
        etiqueta_s = "Satélite AstroShield" if idx == 0 else None
        ax.scatter([sx], [sy], [sz], color="blue", s=100, marker="^", edgecolors="white", label=etiqueta_s)
        ax.plot([sx, tx], [sy, ty], [sz, tz], color="blue", linestyle=":", linewidth=0.8)
        ax.text(sx * 1.04, sy * 1.04, sz * 1.04, f"S{idx + 1}", fontsize=8, fontweight="bold", color="navy")
    ax.set_title(
        f"{titulo}\nFitness: {resultado['fitness']:.1f} | "
        f"Cubiertas: {len(cubiertas)}/{len(regiones)} | Válida: {resultado['es_valida']}",
        fontsize=10
    )
    ax.set_xlim([-1.3, 1.3])
    ax.set_ylim([-1.3, 1.3])
    ax.set_zlim([-1.3, 1.3])
    ax.set_box_aspect([1, 1, 1])
    ax.legend(loc="upper left", fontsize=7)
def mostrar_comparacion_3d(config_inicial, config_final, regiones=None, zonas_desechos=None, objetos_ocupados=None):
    """
    Abre una ventana interactiva 3D que permite rotar y comparar
    la configuración inicial (Generación 0) contra la configuración final evolucionada.
    """
    if regiones is None:
        regiones = obtener_regiones_iniciales()
    if zonas_desechos is None:
        zonas_desechos = obtener_zonas_desechos_iniciales()
    if objetos_ocupados is None:
        objetos_ocupados = obtener_objetos_ocupados_iniciales()
    figura = plt.figure(figsize=(15, 7))
    figura.suptitle("AstroShield — Visualización 3D de la Constelación de 8 Satélites", fontsize=13, fontweight="bold")
    ax1 = figura.add_subplot(1, 2, 1, projection="3d")
    dibujar_escenario_en_subgrafico(
        ax1, config_inicial, "Configuración Inicial (Generación 0)",
        regiones, zonas_desechos, objetos_ocupados
    )
    ax2 = figura.add_subplot(1, 2, 2, projection="3d")
    dibujar_escenario_en_subgrafico(
        ax2, config_final, "Configuración Final Evolucionada",
        regiones, zonas_desechos, objetos_ocupados
    )
    plt.tight_layout()
    plt.savefig("constelacion_3d_comparacion.png", dpi=150)
    print("\n[Visualización 3D] Imagen guardada como 'constelacion_3d_comparacion.png'.")
    print("[Visualización 3D] Abriendo ventana interactiva 3D (puedes girar los globos con el mouse)...")
    plt.show()
# ==============================================================================
# BLOQUE 5 (Daisy): pruebas.py
# Script de validación que ejecuta automáticamente los casos solicitados:
# escenario inicial, evolución, zonas restringidas, espacio reducido/inviable,
# variación entre ejecuciones y prueba adicional propia.
# ==============================================================================
def imprimir_separador(titulo):
    print("\n" + "=" * 78)
    print(f"  {titulo}")
    print("=" * 78)
def prueba_1_escenario_inicial_y_evolucion():
    """
    Prueba 1 y 2: Evalúa el escenario inicial (Generación 0) y analiza cómo
    cambia la calidad de las configuraciones a lo largo de las generaciones.
    """
    imprimir_separador("PRUEBA 1 Y 2: ESCENARIO INICIAL Y EVOLUCIÓN A LO LARGO DE LAS GENERACIONES")
    random.seed(42)
    resultados = ejecutar_algoritmo_evolutivo(
        tamano_poblacion=60,
        generaciones=100,
        probabilidad_mutacion=0.25,
        mostrar_progreso=False
    )
    eval_ini = evaluar_configuracion(resultados["configuracion_inicial"])
    eval_fin = evaluar_configuracion(resultados["configuracion_final"])
    print("[Condiciones]: 8 satélites, 6 regiones oficiales, 3 zonas de desechos, 2 objetos ocupados.")
    print("\n--- Resultados en Generación 0 (Inicial) ---")
    print(f"  Fitness Inicial        : {eval_ini['fitness']:.2f}")
    print(f"  ¿Configuración válida? : {eval_ini['es_valida']}")
    print(f"  Regiones cubiertas ({len(eval_ini['regiones_cubiertas'])}/6): {eval_ini['regiones_cubiertas']}")
    print("\n--- Progreso del Mejor Fitness por Generación ---")
    for gen_muestra in [0, 10, 25, 50, 75, 99]:
        mejor_g = resultados["historial_mejor"][gen_muestra]
        prom_g = resultados["historial_promedio"][gen_muestra]
        print(f"  Generación {gen_muestra:2d} -> Mejor Fitness: {mejor_g:7.2f} | Fitness Promedio Población: {prom_g:7.2f}")
    print("\n--- Resultados en Generación 100 (Final) ---")
    print(f"  Fitness Final          : {eval_fin['fitness']:.2f}")
    print(f"  ¿Configuración válida? : {eval_fin['es_valida']}")
    print(f"  Regiones cubiertas ({len(eval_fin['regiones_cubiertas'])}/6): {eval_fin['regiones_cubiertas']}")
    print("  Distancias mínimas desde un satélite al centro de cada región (km):")
    for nombre_reg, dist in eval_fin["distancias_km"].items():
        print(f"    - {nombre_reg:32s}: {dist:6.2f} km")
    print("\n  Coordenadas finales de los 8 satélites:")
    for idx, sat in enumerate(resultados["configuracion_final"]):
        print(f"    Satélite {idx + 1}: Latitud = {sat[0]:6.2f}°, Longitud = {sat[1]:7.2f}°")
    print("\n[Interpretación]: El algoritmo incrementa el fitness desde un estado inicial con baja cobertura")
    print("hasta lograr cubrir las 6 regiones respetando el 100% de las restricciones espaciales.")
    return resultados
def prueba_2_validacion_zonas_restringidas_y_ocupadas(configuracion_final):
    """
    Prueba 3 y 4: Verifica explícitamente que la solución respeta las 3 zonas de desechos
    y las 2 posiciones ocupadas, y comprueba que el filtro detecta violaciones deliberadas.
    """
    imprimir_separador("PRUEBA 3 Y 4: VALIDACIÓN DE ZONAS DE DESECHOS Y POSICIONES OCUPADAS")
    zonas = obtener_zonas_desechos_iniciales()
    objetos = obtener_objetos_ocupados_iniciales()
    # Parte A: Verificar la solución final obtenida
    violaciones_solucion = verificar_restricciones(configuracion_final, zonas, objetos)
    print("1. Verificación de la solución final evolucionada:")
    print(f"   Violaciones encontradas: {len(violaciones_solucion)}")
    # Parte B: Colocar deliberadamente un satélite en una zona de desechos y otro en el Objeto A
    config_invalida = copy.deepcopy(configuracion_final)
    config_invalida[0] = [45.0, -35.0]  # Exactamente en el centro de Desechos Atlántico Norte
    config_invalida[1] = [10.0, 45.0]   # Exactamente en la posición del Objeto A
    eval_invalida = evaluar_configuracion(config_invalida, None, zonas, objetos)
    print("\n2. Prueba de estrés con satélites forzados dentro de zonas prohibidas:")
    print(f"   ¿Es válida la configuración forzada?: {eval_invalida['es_valida']}")
    print(f"   Fitness tras penalización           : {eval_invalida['fitness']:.2f}")
    print("   Detalle de infracciones detectadas por el filtro:")
    for v in eval_invalida["violaciones"]:
        print(f"     * {v}")
    print("\n[Interpretación]: El filtro detecta inmediatamente cualquier invasión a zonas de desechos")
    print("o posiciones ocupadas y aplica una fuerte penalización (-1000 c/u) que impide su selección.")
def prueba_3_espacio_reducido_y_condicion_inviable():
    """
    Prueba 5 y 6: Evalúa qué ocurre cuando se reduce drásticamente el espacio disponible
    y cuando una condición hace imposible cubrir todas las regiones (escenario inviable).
    """
    imprimir_separador("PRUEBA 5 Y 6: ESPACIO REDUCIDO Y CONDICIÓN DIFÍCIL / INVIABLE")
    random.seed(99)
    # Caso A: Espacio reducido (agregamos 3 nuevas zonas grandes de desechos cerca de regiones)
    zonas_reducidas = obtener_zonas_desechos_iniciales() + [
        {"nombre": "Desechos Caribe Oriental", "lat": 18.0, "lon": -72.0, "radio_grados": 12.0},
        {"nombre": "Desechos Pacífico Norte", "lat": 35.0, "lon": -135.0, "radio_grados": 12.0},
        {"nombre": "Desechos Mediterráneo/Sahara", "lat": 25.0, "lon": 5.0, "radio_grados": 9.0}
    ]
    res_reducido = ejecutar_algoritmo_evolutivo(
        tamano_poblacion=60,
        generaciones=100,
        zonas_desechos=zonas_reducidas,
        mostrar_progreso=False
    )
    eval_red = evaluar_configuracion(res_reducido["configuracion_final"], None, zonas_reducidas, None)
    print("--- Caso A: Espacio Reducido (6 zonas de desechos activas) ---")
    print(f"  ¿Solución válida (sin invadir zonas)?: {eval_red['es_valida']}")
    print(f"  Fitness alcanzado                    : {eval_red['fitness']:.2f}")
    print(f"  Regiones cubiertas ({len(eval_red['regiones_cubiertas'])}/6)           : {eval_red['regiones_cubiertas']}")
    # Caso B: Condición inviable (una tormenta de desechos de 25° de radio bloquea totalmente Yucatán y Amazonía)
    zonas_inviables = obtener_zonas_desechos_iniciales() + [
        {"nombre": "Bloqueo Total Yucatán", "lat": 20.5, "lon": -89.0, "radio_grados": 25.0},
        {"nombre": "Bloqueo Total Amazonía", "lat": -3.5, "lon": -62.0, "radio_grados": 25.0}
    ]
    res_inviable = ejecutar_algoritmo_evolutivo(
        tamano_poblacion=60,
        generaciones=100,
        zonas_desechos=zonas_inviables,
        mostrar_progreso=False
    )
    eval_inv = evaluar_configuracion(res_inviable["configuracion_final"], None, zonas_inviables, None)
    print("\n--- Caso B: Condición Inviable (Yucatán y Amazonía bloqueadas por radio de 25° ~ 2780 km) ---")
    print(f"  ¿Solución válida (respeta bloqueos)? : {eval_inv['es_valida']}")
    print(f"  Fitness alcanzado                    : {eval_inv['fitness']:.2f}")
    print(f"  Regiones cubiertas ({len(eval_inv['regiones_cubiertas'])}/6)           : {eval_inv['regiones_cubiertas']}")
    print(f"  Regiones sacrificadas por seguridad  : {eval_inv['regiones_no_cubiertas']}")
    print("\n[Interpretación]: Cuando es físicamente imposible cubrir una región sin entrar en una zona")
    print("restringida, el algoritmo prefiere mantener la seguridad de la constelación (0 violaciones)")
    print("y maximiza la cobertura de las 4 regiones restantes disponibles.")
def prueba_4_variacion_entre_ejecuciones():
    """
    Prueba 7: Ejecuta el algoritmo 5 veces con distintas semillas aleatorias para analizar
    la estabilidad y variación entre ejecuciones bajo condiciones comparables.
    """
    imprimir_separador("PRUEBA 7: VARIACIÓN ENTRE MÚLTIPLES EJECUCIONES INDEPENDIENTES")
    semillas = [10, 20, 30, 40, 50]
    puntajes = []
    for idx, sem in enumerate(semillas):
        random.seed(sem)
        res = ejecutar_algoritmo_evolutivo(
            tamano_poblacion=60,
            generaciones=100,
            mostrar_progreso=False
        )
        ev = evaluar_configuracion(res["configuracion_final"])
        puntajes.append(ev["fitness"])
        print(
            f"  Ejecución {idx + 1} (Semilla {sem}) -> Fitness: {ev['fitness']:7.2f} | "
            f"Regiones cubiertas: {len(ev['regiones_cubiertas'])}/6 | Válida: {ev['es_valida']}"
        )
    promedio_ejecuciones = sum(puntajes) / len(puntajes)
    diferencia_max_min = max(puntajes) - min(puntajes)
    print(f"\n  Fitness Promedio de las 5 ejecuciones : {promedio_ejecuciones:.2f}")
    print(f"  Rango de variación (Máx - Mín)        : {diferencia_max_min:.2f}")
    print("\n[Interpretación]: En todas las ejecuciones se cubren las 6 regiones sin violar restricciones;")
    print("las pequeñas diferencias de puntaje se deben a la posición exacta de los 2 satélites de respaldo.")
def prueba_5_adicional_cambio_de_prioridades():
    """
    Prueba 8 (Adicional diseñada por el equipo): Evalúa la adaptación del sistema cuando la
    organización cambia dinámicamente las prioridades y agrega una séptima región de emergencia.
    """
    imprimir_separador("PRUEBA 8 (ADICIONAL PROPIA): CAMBIO DINÁMICO DE PRIORIDADES Y NUEVA REGIÓN")
    random.seed(77)
    regiones_modificadas = obtener_regiones_iniciales()
    # Australia pasa de prioridad Normal (100) a Prioridad Crítica (500 puntos)
    for r in regiones_modificadas:
        if "Australia" in r["nombre"]:
            r["prioridad"] = "Crítica"
            r["puntos"] = 500
    # Se agrega una 7ma región de observación en Europa Central
    regiones_modificadas.append(
        {"nombre": "Europa Central", "lat": 48.0, "lon": 10.0, "prioridad": "Alta", "puntos": 300}
    )
    res_cambio = ejecutar_algoritmo_evolutivo(
        tamano_poblacion=60,
        generaciones=100,
        regiones=regiones_modificadas,
        mostrar_progreso=False
    )
    ev_cambio = evaluar_configuracion(res_cambio["configuracion_final"], regiones=regiones_modificadas)
    print(f"  Regiones cubiertas ({len(ev_cambio['regiones_cubiertas'])}/7): {ev_cambio['regiones_cubiertas']}")
    print(f"  Distancia del satélite más cercano a Interior de Australia: {ev_cambio['distancias_km']['Interior de Australia']} km")
    print(f"  Fitness total con nuevas prioridades: {ev_cambio['fitness']:.2f} | Válida: {ev_cambio['es_valida']}")
    print("\n[Interpretación]: El sistema se adapta sin modificar el código base, reasignando uno de los")
    print("satélites excedentes para cubrir la nueva región en Europa y priorizando Australia.")
# ==============================================================================
# PUNTO DE ENTRADA PRINCIPAL (EJECUCIÓN COMPLETA: PRUEBAS + GLOBO 3D)
# (NOMBREEEE)
# ==============================================================================
if __name__ == "__main__":
    # 1. Ejecutar todas las pruebas de validación y mostrar resultados en consola
    resultados_principales = prueba_1_escenario_inicial_y_evolucion()
    prueba_2_validacion_zonas_restringidas_y_ocupadas(resultados_principales["configuracion_final"])
    prueba_3_espacio_reducido_y_condicion_inviable()
    prueba_4_variacion_entre_ejecuciones()
    prueba_5_adicional_cambio_de_prioridades()
    print("\n" + "=" * 78)
    print("  TODAS LAS PRUEBAS FINALIZARON EXITOSAMENTE - ABRIENDO GLOBO 3D")
    print("=" * 78)
    # 2. Mostrar el globo terrestre interactivo en 3D (Generación 0 vs Generación 100)
    mostrar_comparacion_3d(
        resultados_principales["configuracion_inicial"],
        resultados_principales["configuracion_final"]
    )
```
