```python
# PROYECTO: AstroShield — Evolucionar una constelación inteligente
import math
import random
import copy
import numpy as np
import matplotlib.pyplot as plt
# ==============================================================================
# BLOQUE 3 (COMMIT 3): restricciones_fitness.py
# Contiene la función de fitness, el cálculo de cobertura de las 6 regiones
# con sus prioridades, y el filtro de las 3 zonas de desechos y 2 objetos
#Elaborado por Iñaki Villalobos
# ==============================================================================

RADIO_TIERRA_KM = 6371.0
RADIO_OBSERVACION_REGION_KM = 300.0
ALCANCE_MAXIMO_SATELITE_KM = 1200.0  # Radio dentro del cual el satélite logra cubrir los 300 km de la región


def obtener_regiones_iniciales():
    """Devuelve las 6 regiones de interés con sus coordenadas y prioridades."""
    return [
        {"nombre": "Península de Yucatán, México", "lat": 20.5, "lon": -89.0, "prioridad": "Alta", "puntos": 300},
        {"nombre": "Amazonía brasileña", "lat": -3.5, "lon": -62.0, "prioridad": "Alta", "puntos": 300},
        {"nombre": "California, Estados Unidos", "lat": 36.8, "lon": -119.4, "prioridad": "Media", "puntos": 200},
        {"nombre": "Región del Sahel, África", "lat": 15.0, "lon": 0.0, "prioridad": "Media", "puntos": 200},
        {"nombre": "Japón", "lat": 36.0, "lon": 138.0, "prioridad": "Media", "puntos": 200},
        {"nombre": "Interior de Australia", "lat": -25.0, "lon": 133.0, "prioridad": "Normal", "puntos": 100}
    ]


def obtener_zonas_desechos_iniciales():
    """Devuelve las 3 zonas restringidas por desechos espaciales (radios en grados)."""
    return [
        {"nombre": "Atlántico Norte", "lat": 45.0, "lon": -35.0, "radio_grados": 12.0},
        {"nombre": "Pacífico Sur", "lat": -30.0, "lon": -120.0, "radio_grados": 10.0},
        {"nombre": "Región polar norte", "lat": 78.0, "lon": 30.0, "radio_grados": 8.0}
    ]


def obtener_objetos_ocupados_iniciales():
    """Devuelve los 2 objetos existentes con su radio de exclusión en grados."""
    return [
        {"nombre": "Objeto A", "lat": 10.0, "lon": 45.0, "radio_grados": 5.0},
        {"nombre": "Objeto B", "lat": -15.0, "lon": 110.0, "radio_grados": 5.0}
    ]


def calcular_distancia_angular_grados(lat1, lon1, lat2, lon2):
    """
    Calcula la distancia angular en grados sobre una esfera entre dos coordenadas
    usando la fórmula esférica de Haversine.
    """
    rad_lat1 = math.radians(lat1)
    rad_lon1 = math.radians(lon1)
    rad_lat2 = math.radians(lat2)
    rad_lon2 = math.radians(lon2)

    delta_lat = rad_lat2 - rad_lat1
    delta_lon = rad_lon2 - rad_lon1

    a = (math.sin(delta_lat / 2.0) ** 2) + \
        math.cos(rad_lat1) * math.cos(rad_lat2) * (math.sin(delta_lon / 2.0) ** 2)

    if a > 1.0:
        a = 1.0
    if a < 0.0:
        a = 0.0

    angulo_central_rad = 2.0 * math.atan2(math.sqrt(a), math.sqrt(1.0 - a))
    distancia_grados = math.degrees(angulo_central_rad)
    return distancia_grados


def calcular_distancia_km(lat1, lon1, lat2, lon2):
    """Calcula la distancia superficial en kilómetros entre dos puntos de la Tierra."""
    grados = calcular_distancia_angular_grados(lat1, lon1, lat2, lon2)
    radianes = math.radians(grados)
    return RADIO_TIERRA_KM * radianes


def verificar_restricciones(configuracion, zonas_desechos, objetos_ocupados):
    """
    Revisa si algún satélite invade las 3 zonas de desechos, los 2 objetos ocupados,
    o si dos satélites chocan entre sí.
    """
    violaciones = []

    for i in range(len(configuracion)):
        sat_lat = configuracion[i][0]
        sat_lon = configuracion[i][1]

        # 1. Revisar las 3 zonas de desechos espaciales
        for zona in zonas_desechos:
            dist_grados = calcular_distancia_angular_grados(
                sat_lat, sat_lon, zona["lat"], zona["lon"]
            )
            if dist_grados <= zona["radio_grados"]:
                violaciones.append(
                    f"Satélite {i + 1} invade zona de desechos '{zona['nombre']}' "
                    f"(distancia {dist_grados:.2f}° <= {zona['radio_grados']}°)"
                )

        # 2. Revisar los 2 objetos existentes
        for obj in objetos_ocupados:
            dist_grados = calcular_distancia_angular_grados(
                sat_lat, sat_lon, obj["lat"], obj["lon"]
            )
            if dist_grados <= obj["radio_grados"]:
                violaciones.append(
                    f"Satélite {i + 1} invade zona de exclusión de '{obj['nombre']}' "
                    f"(distancia {dist_grados:.2f}° <= {obj['radio_grados']}°)"
                )

        # 3. Revisar separación mínima entre satélites de la misma constelación (mínimo 2°)
        for j in range(i + 1, len(configuracion)):
            dist_entre_sat = calcular_distancia_angular_grados(
                sat_lat, sat_lon, configuracion[j][0], configuracion[j][1]
            )
            if dist_entre_sat < 2.0:
                violaciones.append(
                    f"Satélites {i + 1} y {j + 1} demasiado cerca ({dist_entre_sat:.2f}° < 2.0°)"
                )

    return violaciones


def evaluar_configuracion(configuracion, regiones=None, zonas_desechos=None, objetos_ocupados=None):
    """
    Calcula el fitness total de una constelación de 8 satélites, considerando:
    - Puntos por cubrir cada una de las 6 regiones según su prioridad.
    - Bono por cercanía óptima dentro del radio de 300 km de cada región.
    - Penalización fuerte por cada violación a zonas de desechos u objetos existentes.
    """
    if regiones is None:
        regiones = obtener_regiones_iniciales()
    if zonas_desechos is None:
        zonas_desechos = obtener_zonas_desechos_iniciales()
    if objetos_ocupados is None:
        objetos_ocupados = obtener_objetos_ocupados_iniciales()

    puntaje_total = 0.0
    regiones_cubiertas = []
    regiones_no_cubiertas = []
    distancias_minimas_regiones = {}

    # 1. Evaluar cobertura de las 6 regiones
    for region in regiones:
        mejor_distancia_km = 999999.0

        for satelite in configuracion:
            dist_km = calcular_distancia_km(
                satelite[0], satelite[1], region["lat"], region["lon"]
            )
            if dist_km < mejor_distancia_km:
                mejor_distancia_km = dist_km

        distancias_minimas_regiones[region["nombre"]] = round(mejor_distancia_km, 2)

        if mejor_distancia_km <= ALCANCE_MAXIMO_SATELITE_KM:
            regiones_cubiertas.append(region["nombre"])
            puntaje_total += region["puntos"]

            # Bono extra si el satélite se posiciona dentro de los 300 km centrales de observación
            if mejor_distancia_km <= RADIO_OBSERVACION_REGION_KM:
                bono_precision = 50.0 * (1.0 - (mejor_distancia_km / RADIO_OBSERVACION_REGION_KM))
                puntaje_total += bono_precision
            else:
                proporcion = (ALCANCE_MAXIMO_SATELITE_KM - mejor_distancia_km) / (
                    ALCANCE_MAXIMO_SATELITE_KM - RADIO_OBSERVACION_REGION_KM
                )
                puntaje_total += 20.0 * proporcion
        else:
            regiones_no_cubiertas.append(region["nombre"])
            guia_cercania = max(0.0, 20.0 * (1.0 - (mejor_distancia_km / 10000.0)))
            puntaje_total += guia_cercania

    # 2. Bono por buena distribución entre satélites (evitar amontonamiento innecesario)
    for i in range(len(configuracion)):
        for j in range(i + 1, len(configuracion)):
            dist_grados = calcular_distancia_angular_grados(
                configuracion[i][0], configuracion[i][1],
                configuracion[j][0], configuracion[j][1]
            )
            if dist_grados >= 10.0:
                puntaje_total += 1.5

    # 3. Aplicar filtro de restricciones (zonas de desechos y posiciones ocupadas)
    violaciones = verificar_restricciones(configuracion, zonas_desechos, objetos_ocupados)
    es_valida = len(violaciones) == 0

    if not es_valida:
        puntaje_total -= 1000.0 * len(violaciones)

    return {
        "fitness": puntaje_total,
        "es_valida": es_valida,
        "violaciones": violaciones,
        "regiones_cubiertas": regiones_cubiertas,
        "regiones_no_cubiertas": regiones_no_cubiertas,
        "distancias_km": distancias_minimas_regiones
    }


# ==============================================================================
# BLOQUE 2 (Venus): algoritmo_evolutivo.py
# Contiene la representación de los 8 satélites y la estructura principal
# del algoritmo evolutivo (generación, selección por torneo, cruza y mutación).
# ==============================================================================
NUM_SATELITES = 8
def crear_satelite_aleatorio():
    """Crea un satélite con latitud [-90, 90] y longitud [-180, 180] aleatorias."""
    latitud = random.uniform(-90.0, 90.0)
    longitud = random.uniform(-180.0, 180.0)
    return [latitud, longitud]
def crear_configuracion_aleatoria():
    """Genera una constelación de 8 satélites en posiciones aleatorias."""
    configuracion = []
    for i in range(NUM_SATELITES):
        satelite = crear_satelite_aleatorio()
        configuracion.append(satelite)
    return configuracion
def crear_poblacion_inicial(tamano_poblacion):
    """Crea una lista de configuraciones aleatorias."""
    poblacion = []
    for i in range(tamano_poblacion):
        individuo = crear_configuracion_aleatoria()
        poblacion.append(individuo)
    return poblacion
def seleccion_por_torneo(poblacion, puntajes_fitness, tamano_torneo=3):
    """Selecciona al mejor individuo de una muestra aleatoria de la población."""
    indices_participantes = []
    total_individuos = len(poblacion)
    for i in range(tamano_torneo):
        indice_azar = random.randint(0, total_individuos - 1)
        indices_participantes.append(indice_azar)
    mejor_indice = indices_participantes[0]
    mejor_puntaje = puntajes_fitness[mejor_indice]
    for indice in indices_participantes:
        if puntajes_fitness[indice] > mejor_puntaje:
            mejor_puntaje = puntajes_fitness[indice]
            mejor_indice = indice
    return copy.deepcopy(poblacion[mejor_indice])
def cruzar_configuraciones(padre1, padre2):
    """Cruza dos configuraciones de 8 satélites usando un punto de corte."""
    punto_corte = random.randint(1, NUM_SATELITES - 1)
    hijo = []
    for i in range(NUM_SATELITES):
        if i < punto_corte:
            hijo.append(copy.deepcopy(padre1[i]))
        else:
            hijo.append(copy.deepcopy(padre2[i]))
    return hijo
def mutar_configuracion(configuracion, probabilidad_mutacion=0.25):
    """Modifica ligeramente las coordenadas de los satélites según una probabilidad."""
    nueva_config = copy.deepcopy(configuracion)
    for i in range(NUM_SATELITES):
        azar = random.random()
        if azar < probabilidad_mutacion:
            # 80% de las veces hacemos un ajuste fino Gaussiano
            # 20% de las veces movemos el satélite a una posición totalmente nueva
            if random.random() < 0.8:
                cambio_lat = random.gauss(0, 8.0)
                cambio_lon = random.gauss(0, 8.0)
                nueva_lat = nueva_config[i][0] + cambio_lat
                nueva_lon = nueva_config[i][1] + cambio_lon
                if nueva_lat > 90.0:
                    nueva_lat = 90.0
                if nueva_lat < -90.0:
                    nueva_lat = -90.0
                if nueva_lon > 180.0:
                    nueva_lon = nueva_lon - 360.0
                if nueva_lon < -180.0:
                    nueva_lon = nueva_lon + 360.0
                nueva_config[i] = [nueva_lat, nueva_lon]
            else:
                nueva_config[i] = crear_satelite_aleatorio()
    return nueva_config
def ejecutar_algoritmo_evolutivo(
    tamano_poblacion=60,
    generaciones=100,
    probabilidad_mutacion=0.25,
    regiones=None,
    zonas_desechos=None,
    objetos_ocupados=None,
    mostrar_progreso=True
):
    """
    Ejecuta el ciclo principal del algoritmo evolutivo y devuelve:
    - configuracion_inicial (mejor de la generación 0)
    - configuracion_final (mejor global encontrada)
    - historial_mejor
    - historial_promedio
    """
    if regiones is None:
        regiones = obtener_regiones_iniciales()
    if zonas_desechos is None:
        zonas_desechos = obtener_zonas_desechos_iniciales()
    if objetos_ocupados is None:
        objetos_ocupados = obtener_objetos_ocupados_iniciales()
    poblacion = crear_poblacion_inicial(tamano_poblacion)
    historial_mejor_fitness = []
    historial_promedio_fitness = []
    mejor_configuracion_inicial = None
    mejor_configuracion_global = None
    mejor_fitness_global = -999999.0
    for gen in range(generaciones):
        puntajes_fitness = []
        suma_fitness = 0.0
        # 1. Evaluar toda la población actual
        for individuo in poblacion:
            resultado = evaluar_configuracion(
                individuo, regiones, zonas_desechos, objetos_ocupados
            )
            fitness_ind = resultado["fitness"]
            puntajes_fitness.append(fitness_ind)
            suma_fitness += fitness_ind
            if fitness_ind > mejor_fitness_global:
                mejor_fitness_global = fitness_ind
                mejor_configuracion_global = copy.deepcopy(individuo)
        # Guardar la mejor configuración de la generación 0 para comparar después
        if gen == 0:
            indice_mejor_gen0 = 0
            for i in range(len(puntajes_fitness)):
                if puntajes_fitness[i] > puntajes_fitness[indice_mejor_gen0]:
                    indice_mejor_gen0 = i
            mejor_configuracion_inicial = copy.deepcopy(poblacion[indice_mejor_gen0])
        promedio_actual = suma_fitness / tamano_poblacion
        historial_mejor_fitness.append(mejor_fitness_global)
        historial_promedio_fitness.append(promedio_actual)
        if mostrar_progreso and (gen % 20 == 0 or gen == generaciones - 1):
            info_mejor = evaluar_configuracion(
                mejor_configuracion_global, regiones, zonas_desechos, objetos_ocupados
            )
            print(
                f"Generación {gen:3d} | Mejor Fitness: {mejor_fitness_global:7.2f} | "
                f"Promedio: {promedio_actual:7.2f} | "
                f"Regiones cubiertas: {len(info_mejor['regiones_cubiertas'])}/{len(regiones)} | "
                f"Válida: {info_mejor['es_valida']}"
            )
        # 2. Ordenar índices de mejor a peor para aplicar Elitismo (conservar los 2 mejores)
        indices_ordenados = list(range(tamano_poblacion))
        for i in range(tamano_poblacion):
            for j in range(i + 1, tamano_poblacion):
                if puntajes_fitness[indices_ordenados[j]] > puntajes_fitness[indices_ordenados[i]]:
                    temp = indices_ordenados[i]
                    indices_ordenados[i] = indices_ordenados[j]
                    indices_ordenados[j] = temp
        nueva_poblacion = []
        nueva_poblacion.append(copy.deepcopy(poblacion[indices_ordenados[0]]))
        nueva_poblacion.append(copy.deepcopy(poblacion[indices_ordenados[1]]))
        # 3. Generar el resto de la nueva población mediante torneo, cruza y mutación
        while len(nueva_poblacion) < tamano_poblacion:
            padre1 = seleccion_por_torneo(poblacion, puntajes_fitness)
            padre2 = seleccion_por_torneo(poblacion, puntajes_fitness)
            hijo = cruzar_configuraciones(padre1, padre2)
            hijo_mutado = mutar_configuracion(hijo, probabilidad_mutacion)
            nueva_poblacion.append(hijo_mutado)
        poblacion = nueva_poblacion
    return {
        "configuracion_inicial": mejor_configuracion_inicial,
        "configuracion_final": mejor_configuracion_global,
        "historial_mejor": historial_mejor_fitness,
        "historial_promedio": historial_promedio_fitness
    }
```
